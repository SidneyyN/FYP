```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def plot_experiment_results(path_exp, runs):
    colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k']  # Basic color codes for matplotlib
    plt.figure(figsize=(10, 6))

    for i, run in enumerate(runs):
        filename = f'experiment_results_run_{run}.csv'
        filepath = os.path.join(path_exp, filename)

        if os.path.exists(filepath):
            df = pd.read_csv(filepath)
            plt.plot(df.index, df['Predicted Price'], linestyle='--', color=colors[i % len(colors)], label=f'Run {run} - Predicted Price')
            plt.plot(df.index, df['Actual Price'], linestyle='-', color=colors[i % len(colors)], label=f'Run {run} - Actual Price')
        else:
            print(f'File not found: {filename}')

    plt.title('Experiment Results')
    plt.xlabel('Time Step')
    plt.ylabel('Price')
    plt.legend()
    plt.grid(True)
    plt.savefig('../results/fig/prices_neg.png')
    plt.show()

# Specify the path to the experiment results and the runs to plot
path_exp = '../results/experiments/'
runs_to_plot = [0, 1, 2, 3, 4]
plot_experiment_results(path_exp, runs_to_plot)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def plot_experiment_results(path_exp):
    files = [f for f in os.listdir(path_exp) if f.endswith('.csv')]

    for file in files:
        filepath = os.path.join(path_exp, file)
        df = pd.read_csv(filepath)

        plt.figure(figsize=(10, 6))
        plt.plot(df.index, df['Predicted Price'], label='Predicted Price (pe_list)')
        plt.plot(df.index, df['Actual Price'], label='Actual Price (p_list)')
        plt.title(f'Experiment Results: {file}')
        plt.xlabel('Time Step')
        plt.ylabel('Price')
        plt.legend()
        plt.grid(True)
        plt.show()

# Specify the path to the experiment results
path_exp = '../results/experiments/'
plot_experiment_results(path_exp)
```
