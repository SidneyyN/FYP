```python
import pandas as pd
import statsmodels.api as sm
```

## Positive Feedback

### High-Volatility Setup (WITH CONSTRAINT)

```python
# Step 1: Load the Excel file
# Replace 'your_file.xlsx' with the actual filename of your Excel file
file_path = "../results/experiments/experiment_batch_20250308_153210_Pos_High-Vol_Constraint/consolidated_results.csv"  # Update this with your file path
data = pd.read_csv(file_path)
```

```python
# Step 2: Aggregate data by time step to calculate mean predicted and actual prices
aggregated_data = data.groupby('time_step').agg({
    'mean_predicted_price': 'mean',      # Mean predicted price
    'mean_actual_price': 'mean'    # Mean actual price
}).reset_index()

# Step 3: Create lagged variables
aggregated_data['p_t-1'] = aggregated_data['mean_actual_price'].shift(1)
aggregated_data['p_t-2'] = aggregated_data['mean_actual_price'].shift(2)
aggregated_data['p_e,h,t-1'] = aggregated_data['mean_predicted_price'].shift(1)

# Step 4: Calculate price difference
aggregated_data['price_diff'] = aggregated_data['p_t-1'] - aggregated_data['p_t-2']

# Step 5: Drop rows with missing values due to lagging
aggregated_data = aggregated_data.dropna()

# Step 6: Prepare regression variables
X = aggregated_data[['p_t-1', 'p_e,h,t-1', 'price_diff']]  # Independent variables
X['constant'] = 60  # Add the constant term (60)
y = aggregated_data['mean_predicted_price']  # Dependent variable (mean predicted price)

# Step 7: Add intercept for regression
X = sm.add_constant(X)

# Just to make sur that all columns in X are numeric 
X = X.apply(pd.to_numeric, errors='coerce')

# Ensure y is numeric as well
y = pd.to_numeric(y, errors='coerce')

# Step 8: Run the regression
model = sm.OLS(y, X).fit()
```

```python
# Step 9: Output the regression results
print(model.summary())
```
