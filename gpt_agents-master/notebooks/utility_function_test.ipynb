## Debugging on why wrong values

```python
import os

csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]
print(f"Total CSV files detected: {len(csv_files)}")
print("Files:", csv_files)
```

```python
# Normalize the folder path to avoid mix of slashes
folder_path = os.path.abspath(folder_path)
```

```python
import pandas as pd

all_time_steps = set()

for file in csv_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_csv(file_path)

    # Debugging step: Print unique time_step values in each file
    print(f"{file} - Unique time_steps:", df["time_step"].unique())

    all_time_steps.update(df["time_step"].unique())

print("\nTotal unique time steps across all files:", sorted(all_time_steps))
```

```python
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_csv(file_path)

    print(f"\nChecking {file} - Data Types:")
    print(df.dtypes)  # Should show 'float' for predicted_price & actual_price
```

```python
import pandas as pd

all_data = []

for file in csv_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_csv(file_path)

    all_data.append(df[["time_step", "predicted_price", "actual_price"]])

# Combine all files into one DataFrame
combined_df = pd.concat(all_data, ignore_index=True)

# Compute manual mean values
manual_summary = combined_df.groupby("time_step").agg(
    manual_mean_predicted_price=("predicted_price", "mean"),
    manual_mean_actual_price=("actual_price", "mean")
).reset_index()

# Print a sample for comparison
print(manual_summary.head())
```

```python
print(combined_df[combined_df["time_step"] == 0])  # Show rows where time_step == 10
```

```python
duplicates = combined_df[combined_df.duplicated(subset=["time_step"], keep=False)]
print("Duplicate time_step values found:")
print(duplicates)
```

```python
# Pick a time step to inspect, e.g., time_step = 10
time_step_to_check = 0

# Extract rows for the selected time step
df_check = combined_df[combined_df["time_step"] == time_step_to_check]

# Manually compute the mean for predicted_price and actual_price
mean_predicted = df_check["predicted_price"].mean()
mean_actual = df_check["actual_price"].mean()

print(f"Time Step {time_step_to_check}")
print(f"Manual Mean Predicted Price: {mean_predicted}")
print(f"Manual Mean Actual Price: {mean_actual}")

# Display all values contributing to this mean
print("\nRaw values for predicted_price:")
print(df_check["predicted_price"].values)

print("\nRaw values for actual_price:")
print(df_check["actual_price"].values)
```
