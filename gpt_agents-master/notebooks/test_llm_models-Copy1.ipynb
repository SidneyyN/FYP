### Test JSON format

```python
llm_35 = "gpt-3.5-turbo-1106"
```

```python
messages = [
        {"role": "system", "content": "You are a kind helpful assistant. Reply to each question in JSON format with key 'reply'"},]

new_messages = ['What is the cutest animal?',\
        'What do they like to eat?']
```

```python
temperature = 1
```

```python
for new_message in new_messages:

    messages.append({"role": "user", "content": new_message})
    chat_completion = client.chat.completions.create(
            model=llm_35, 
            messages=messages,
            temperature=temperature,
            response_format= { 'type': "json_object" }
    )
    reply = chat_completion.choices[0].message.content
    print(reply)
```

```python
messages = [
        {"role": "system", "content": "You are a kind helpful assistant."},]

new_messages = ['What is the cutest animal?',\
        'What do they like to eat?']

for new_message in new_messages:
    # Add the new message to the messages to maintain context
    messages.append({"role": "user", "content": new_message})
    # Concatenate all previous messages and responses to form the context
    prompt_text = '\n'.join([message['content'] for message in messages])

    # Get the completion from the model
    completion = client.completions.create(
        model=llm_da,
        prompt=prompt_text,
        temperature=temperature
    )
    reply = completion.choices[0].text

    # Add the response to the messages to maintain context
    messages.append({"role": "assistant", "content": reply})

    # Print the response
    print(prompt_text)
    print(reply)
   
```
