```python
# set simulation parameters (fixed)
noise_mean = 0
noise_sd = 1/4
instruction_type = 'original_30-50w'
n_agents = 6
n_steps = 50
gpt_model= "gpt-3.5-turbo-1106"#"gpt-4"
```

```python
#grid parameters

# Define the lists
memory_values = [1, 3, 5]
temperature_values = [0.3, 0.7, 1.0]
feedback_values = ['pos', 'neg']
expmnt_nums = [1, 2, 3]

# Generate all combinations
combinations = list(product(memory_values, temperature_values, feedback_values, expmnt_nums))

# Create a DataFrame from the combinations
df = pd.DataFrame(combinations, columns=['memory', 'temperature', 'feedback', 'expmnt_num'])
```

```python
 gpt_model 
```

```python
for i in range(15, df.shape[0]):

    memory = df["memory"].iloc[i]
    temperature = df["temperature"].iloc[i]
    feedback = df["feedback"].iloc[i]
    expmnt_num = df["expmnt_num"].iloc[i]
    
    expmnt_num = check_existing_file(path_exp, expmnt_num, feedback, instruction_type, temperature, memory, n_steps, n_agents, gpt_model)
    seed = expmnt_num
    np.random.seed(seed)

    p, pe, rewards, messages_list  = mod.run_experiment(seed,\
                    expmnt_num, noise_mean, noise_sd, temperature, memory=memory, gpt_model=gpt_model, n_steps=n_steps, n_agents=n_agents,
                    feedback=feedback, instruction_type=instruction_type, prints_on=False,\
                    save_res_bool=True, path_exp=path_exp)
```
